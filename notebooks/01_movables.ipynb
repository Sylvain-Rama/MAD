### Notebook for test and demonstration of Base Movable objects

```python
from mad.objs.base import MovableObj
import numpy as np
```

```python
# Movables are simple points in space that can be moved around. They have a position, a velocity and a name.
# If undeclared, velocity is set to 0 and name is set to "MovableObj". The position is a list of two floats, representing the x and y coordinates of the object in space.
# A minimum of 2 dimensions is required, the 3rd one will be expected to be 0 if not declared.
obj1 = MovableObj(position = [1, 1])
obj2 = MovableObj(position = [0, 1])
```

```python
# Normalize will give the norm of the position
obj1.position, obj1.normalize
```

```python
# We have another convenience method to get the distance between two objects. It is symmetric.
np.isclose(obj1.distance(obj2), obj2.distance(obj1))
```

```python
# And finally, we can check if two objects are equal. This is done by checking if their _id attributes are equal.
obj1 == obj1, obj1 == obj2
```
